```python
#!/usr/bin/env python3
"""
random_compile.py

Build a video by randomly picking images/videos from a directory and
stitching them together, each shown for a fixed (or random) duration.

Usage:
    python random_compile.py -i C:\path\to\clips -o out.mp4 -n 30 -d 1

    -i  input directory containing images and/or videos
    -o  output video file
    -n  number of clips to pick (with replacement, so repeats are possible)
    -d  duration in seconds each clip/image is shown (default 1)
    --min-d / --max-d  instead of -d, randomize duration per clip in a range
    --res  output resolution, default 1280x720
    --fps  output fps, default 30
    --seed optional random seed for reproducible output

Requires ffmpeg and ffprobe on PATH.
"""

import argparse
import os
import random
import subprocess
import sys
import tempfile
import shutil

IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff"}
VIDEO_EXTS = {".mp4", ".mov", ".mkv", ".avi", ".webm", ".m4v"}


def ffprobe_duration(path):
    """Return duration in seconds as float, or None on failure."""
    try:
        out = subprocess.check_output(
            [
                "ffprobe", "-v", "error",
                "-show_entries", "format=duration",
                "-of", "default=noprint_wrappers=1:nokey=1",
                path,
            ],
            stderr=subprocess.STDOUT,
        )
        return float(out.strip())
    except Exception:
        return None


def build_clip(src, dst, duration, res, fps, is_image):
    w, h = res
    vf = (
        f"scale={w}:{h}:force_original_aspect_ratio=increase,"
        f"crop={w}:{h},setsar=1,fps={fps}"
    )

    if is_image:
        cmd = [
            "ffmpeg", "-y", "-loop", "1", "-t", str(duration), "-i", src,
            "-vf", vf,
            "-c:v", "libx264", "-pix_fmt", "yuv420p",
            "-an",
            dst,
        ]
    else:
        total = ffprobe_duration(src)
        if total is None or total <= 0:
            print(f"  skipping (couldn't read duration): {src}")
            return False
        if total <= duration:
            start = 0
        else:
            start = random.uniform(0, total - duration)
        cmd = [
            "ffmpeg", "-y", "-ss", f"{start:.3f}", "-i", src,
            "-t", str(duration),
            "-vf", vf,
            "-c:v", "libx264", "-pix_fmt", "yuv420p",
            "-an",
            dst,
        ]

    result = subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.PIPE)
    if result.returncode != 0:
        print(f"  ffmpeg failed on {src}:")
        print(result.stderr.decode(errors="ignore")[-800:])
        return False
    return True


def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("-i", "--input", required=True, help="source directory")
    ap.add_argument("-o", "--output", required=True, help="output video file")
    ap.add_argument("-n", "--num-clips", type=int, default=20)
    ap.add_argument("-d", "--duration", type=float, default=1.0)
    ap.add_argument("--min-d", type=float, default=None)
    ap.add_argument("--max-d", type=float, default=None)
    ap.add_argument("--res", default="1280x720")
    ap.add_argument("--fps", type=int, default=30)
    ap.add_argument("--seed", type=int, default=None)
    args = ap.parse_args()

    if args.seed is not None:
        random.seed(args.seed)

    w, h = (int(x) for x in args.res.lower().split("x"))

    files = []
    for name in os.listdir(args.input):
        ext = os.path.splitext(name)[1].lower()
        if ext in IMAGE_EXTS or ext in VIDEO_EXTS:
            files.append(os.path.join(args.input, name))

    if not files:
        sys.exit("No images or videos found in that directory.")

    tmpdir = tempfile.mkdtemp(prefix="randcompile_")
    print(f"Working dir: {tmpdir}")

    concat_list_path = os.path.join(tmpdir, "list.txt")
    built = 0

    with open(concat_list_path, "w", encoding="utf-8") as list_f:
        for i in range(args.num_clips):
            src = random.choice(files)
            ext = os.path.splitext(src)[1].lower()
            is_image = ext in IMAGE_EXTS

            if args.min_d is not None and args.max_d is not None:
                dur = round(random.uniform(args.min_d, args.max_d), 2)
            else:
                dur = args.duration

            dst = os.path.join(tmpdir, f"part_{i:04d}.mp4")
            print(f"[{i+1}/{args.num_clips}] {os.path.basename(src)} -> {dur}s")

            if build_clip(src, dst, dur, (w, h), args.fps, is_image):
                # concat demuxer wants forward slashes / escaped paths
                safe_path = dst.replace("\\", "/")
                list_f.write(f"file '{safe_path}'\n")
                built += 1

    if built == 0:
        shutil.rmtree(tmpdir, ignore_errors=True)
        sys.exit("No clips were successfully built.")

    print(f"Concatenating {built} clips...")
    concat_cmd = [
        "ffmpeg", "-y", "-f", "concat", "-safe", "0",
        "-i", concat_list_path,
        "-c:v", "libx264", "-pix_fmt", "yuv420p",
        args.output,
    ]
    result = subprocess.run(concat_cmd, stderr=subprocess.PIPE)
    if result.returncode != 0:
        print(result.stderr.decode(errors="ignore")[-1500:])
        sys.exit("Concat step failed.")

    shutil.rmtree(tmpdir, ignore_errors=True)
    print(f"Done: {args.output}")


if __name__ == "__main__":
    main()
```
