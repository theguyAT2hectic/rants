```python
def convert_webm(input, output):
    from PIL import Image
    # getting the file name and extension
    # Filename and extension extraction is not needed here.


    if output==".jpg":
        im = Image.open(input)
        rgb_im = im.convert('RGB')
        rgb_im.save(output)
    if output==".png":
        im = Image.open(input)
        im.save(output, "PNG")
    if output==".jpeg":
        im = Image.open(input)
        rgb_im = im.convert('RGB')
        rgb_im.save(output, "JPEG")
    if output==".bmp":
        im = Image.open(input)
        im.save(output, "BMP")
    if output==".tiff":
        im = Image.open(input)
        im.save(output, "TIFF")
    if output==".gif":
        im = Image.open(input)
        im.save(output, "GIF")
    if output==".webp":
        im = Image.open(input)
        im.save(output, "WEBP")
    return f"Converted {input} to {output}"

    
```
